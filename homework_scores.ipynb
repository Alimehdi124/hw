```python
math = [80, 92, 75, 60, 88]
physics = [70, 65, 90, 85, 78]
english = [95, 100, 85, 90, 87]

subjects = {"Riyaziyyat": math, "Fizika": physics, "İngilis dili": english}
results = {name: {"average": sum(scores) / len(scores), "minimum": min(scores), "maximum": max(scores)} for name, scores in subjects.items()}
best_subject = max(results, key=lambda name: results[name]["average"])

for name, result in results.items():
    print(name, "| Orta:", result["average"], "| Minimum:", result["minimum"], "| Maksimum:", result["maximum"])
print("Ən yüksək orta balı olan fənn:", best_subject)
```

|---|---:|---:|---:|

| Fizika | 77.6 | 65 | 90 |
| İngilis dili | 91.4 | 85 | 100 |
