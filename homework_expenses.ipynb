```python
expenses = [1200, 1500, 800, 600, 1000, 400, 900]
total = sum(expenses)
average = total / len(expenses)
maximum = max(expenses)
minimum = min(expenses)
days_over_1000 = sum(expense > 1000 for expense in expenses)

print("Ümumi məbləğ:", total)
print("Orta qiymət:", average)
print("Ən yüksək xərc:", maximum)
print("Ən aşağı xərc:", minimum)
print("1000-dən çox olan günlərin sayı:", days_over_1000)
```
