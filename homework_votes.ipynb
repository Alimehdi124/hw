```python
from collections import Counter

votes = ["apple", "banana", "apple", "orange", "banana", "apple", "apple", "orange"]
counts = Counter(votes)
leader = counts.most_common(1)[0][0]
top_three = counts.most_common(3)

print("Səslər:")
for fruit, count in counts.items():
    print(f"{fruit} → {count}")
print("Lider:", leader)
print("İlk 3 nəticə:")
for fruit, count in top_three:
    print(f"{fruit} → {count}")
```
